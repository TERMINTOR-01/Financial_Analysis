# Correlation Analysis

## Objective

The purpose of the correlation analysis is to understand how different numerical variables relate to each other. Correlation values range from **-1 to +1**.

- **+1** → Perfect Positive Correlation
- **0** → No Correlation
- **-1** → Perfect Negative Correlation

Generally,

- **0.70 to 1.00** → Strong Positive
- **0.40 to 0.69** → Moderate Positive
- **0.10 to 0.39** → Weak Positive
- **-0.10 to -0.39** → Weak Negative
- **-0.40 to -0.69** → Moderate Negative
- **-0.70 to -1.00** → Strong Negative

#### 2.Deep vs. Shallow Discount Performance:

```python
df["discount_tier"] = pd.cut(
    df["discount_percentage"],
    bins=[0, 10, 30, 50, 100],
    labels=["0-10%", "10-30%", "30-50%", "50%+"],
    include_lowest=True
)

df["discount_tier"].value_counts()
```

```python
tier_analysis = (
    df.groupby("discount_tier", observed=False)
      .agg(
          Avg_Rating_Count=("rating_count", "mean"),
          Avg_Revenue=("Revenue_Potential", "mean"),
          Avg_Discount=("discount_percentage", "mean"),
          Product_Count=("product_name", "count")
      )
      .reset_index()
)

# Convert to Thousands (K)
tier_analysis["Avg_Revenue"] = (tier_analysis["Avg_Revenue"] / 1000).round(2)

tier_analysis
```

```python
comparison = tier_analysis.melt(
    id_vars="discount_tier",
    value_vars=["Avg_Rating_Count", "Avg_Revenue"],
    var_name="Metric",
    value_name="Value"
)

fig = px.bar(
    comparison,
    x="discount_tier",
    y="Value",
    color="Metric",
    barmode="group",
    title="Discount Tier Performance"
)

fig.update_layout(template="plotly_white")

fig.show()
```

##### 1.0–10% Discount Tier
###### Very low discounts generate reasonable revenue, but customer engagement remains lower than products receiving moderate discounts.

##### 2.10–30% Discount Tier ⭐ (Best Performing)
###### Products with moderate discounts (10–30%) achieve the best balance between customer demand and revenue generation. This appears to be the most effective pricing strategy.

##### 3.30–50% Discount Tier
###### Increasing discounts beyond 30% does not improve customer engagement and is associated with lower revenue potential.

##### 4.50%+ Discount Tier
###### Deep discounts (50%+) do not generate proportionally higher customer engagement and are associated with the lowest revenue potential, suggesting that excessive discounting may be reducing profitability.

#### Discount  Leakage

##### Step 1: Total Discount Leakage

```python
df["Total_Discount_Leakage"] = (
    df["absolute_discounted_amount"] * df["rating_count"]
)
```

```python
category_discount = (
    df.groupby("category")["Total_Discount_Leakage"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

# Convert to Lakhs
category_discount["Total_Discount_Leakage_Lakh"] = (
    category_discount["Total_Discount_Leakage"] / 100000
).round(2)

category_discount
```

```python

fig = px.bar(
    category_discount,
    x="category",
    y="Total_Discount_Leakage",
    title="Total Discount Leakage by Category",
    color="Total_Discount_Leakage",
    text_auto=".2s"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Category",
    yaxis_title="Total Discount Amount (£)"
)

fig.show()
```

##### Step 3: Top 10 Products with Highest Discount Leakage

```python
category_compare = (
    df.groupby("category")
      .agg(
          Total_Discount_Leakage=("Total_Discount_Leakage","sum"),
          Revenue_Potential=("Revenue_Potential","sum")
      )
      .reset_index()
)

comparison = category_compare.melt(
    id_vars="category",
    value_vars=["Total_Discount_Leakage","Revenue_Potential"],
    var_name="Metric",
    value_name="Value"
)

fig = px.bar(
    comparison,
    x="category",
    y="Value",
    color="Metric",
    barmode="group",
    text_auto=".2s"
)

fig.update_layout(height=600, width=1200)
fig.show()
```

# 📊 Price Realization Ratio Analysis

## What is Price Realization Ratio?

The **Price Realization Ratio** measures how much of the original product price the company actually receives after applying discounts.

### Formula

Price Realization Ratio}=Discounted Price/Actual Price


A **higher ratio** indicates that the company preserves more of its original selling price, while a **lower ratio** means the company depends heavily on discounts to drive sales.

---

# What does "Preserve Price Integrity" mean?

**Price Integrity** means customers are willing to purchase products close to their original listed price without requiring large discounts.

### Example

- **Original Price:** £1000
- **Selling Price:** £950

**Price Realization Ratio = 0.95**

This means the company keeps **95% of the original selling price**, indicating strong pricing power and healthy profit margins.

Categories with high Price Realization Ratios generally have:

- Strong customer demand
- Better profit margins
- Lower dependence on discounts
- Higher perceived product value

---

# What does "Rely Heavily on Deep Discounting" mean?

When the Price Realization Ratio is low, the company must significantly reduce prices to generate sales.

### Example

- **Original Price:** £1000
- **Selling Price:** £350

**Price Realization Ratio = 0.35**

This means the company receives only **35% of the original selling price**.

Such categories generally indicate:

- Heavy promotional pricing
- Lower profit margins
- Intense market competition
- Inventory clearance strategies

```python
## Step 1: Sort products by Revenue Potential
pareto_df = (
    df.sort_values("Revenue_Potential", ascending=False)
      .reset_index(drop=True)
)
```

```python
## Step 2: Calculate Total Revenue Potential
total_revenue = pareto_df["Revenue_Potential"].sum()
```

```python
## Step 3: Calculate Cumulative Revenue
pareto_df["Cumulative_Revenue"] = pareto_df["Revenue_Potential"].cumsum()
```

```python
## Step 4: Calculate Cumulative Percentage
pareto_df["Cumulative_%"] = (
    pareto_df["Cumulative_Revenue"] / total_revenue
) * 100
```

```python
## Step 5: Calculate Product Percentage
pareto_df["Product_%"] = (
    (pareto_df.index + 1) / len(pareto_df)
) * 100
```

```python
## Step 6: Find Products Responsible for 80% Revenue
top_products = pareto_df[pareto_df["Cumulative_%"] <= 80]

top_products.head()
```

```python
## Step 7: Top 20 Products
num_products = len(top_products)

percentage_products = (
    num_products / len(df)
) * 100

print(f"{num_products} products generate approximately 80% of Revenue Potential.")
print(f"That is {percentage_products:.2f}% of all products.")
```
